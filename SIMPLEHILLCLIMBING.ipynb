#SIMPLE HILL CLIMBING

```python
# Simple Hill Climbing Algorithm Example
# Problem: Find the maximum value of f(x) = -(x^2) + 10

import random

# Objective function
def objective_function(x):
    return -(x ** 2) + 10


# Generate a neighboring solution
def get_neighbor(x):
    # Move left or right by 1
    step = random.choice([-1, 1])
    return x + step


# Hill Climbing Algorithm
def hill_climbing(start, max_iterations):
    current = start
    current_value = objective_function(current)

    print("Starting at x =", current)
    print("Initial value =", current_value)

    for i in range(max_iterations):

        # Get a nearby solution
        neighbor = get_neighbor(current)
        neighbor_value = objective_function(neighbor)

        print(f"\nIteration {i + 1}")
        print("Neighbor:", neighbor)
        print("Neighbor value:", neighbor_value)

        # Move to neighbor if it is better
        if neighbor_value > current_value:
            current = neighbor
            current_value = neighbor_value
            print("Moved to better neighbor")
        else:
            print("Kept current solution")

    return current, current_value


# Run the algorithm
start_position = random.randint(-10, 10)

best_x, best_value = hill_climbing(start_position, 20)

print("\nBest solution found:")
print("x =", best_x)
print("f(x) =", best_value)
```
